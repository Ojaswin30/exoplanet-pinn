```python
import numpy as np
import matplotlib.pyplot as plt
import lightkurve as lk

search = lk.search_lightcurve("Kepler-10", author="Kepler", cadence="long")
print(search)
lc = search.download_all().stitch().remove_nans()
lc.plot()
plt.show()
```

```python
flat = lc.flatten(window_length=901)
flat.plot()
plt.show()
```

```python
bls = flat.to_periodogram(method="bls", period=np.linspace(0.5, 1.5, 10000), duration=0.07)
bls.plot()
plt.show()
print("Period:", bls.period_at_max_power)
print("Depth:", bls.depth_at_max_power)
```

```python
folded = flat.fold(period=bls.period_at_max_power, epoch_time=bls.transit_time_at_max_power)
ax = folded.scatter(alpha=0.2)
folded.bin(time_bin_size=0.005).plot(ax=ax, c="r", lw=2)
ax.set_xlim(-0.3, 0.3)
plt.show()
```
